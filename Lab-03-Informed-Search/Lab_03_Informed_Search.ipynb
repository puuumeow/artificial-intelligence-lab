```python
print("Laiba Sumaiya Nazim 23549009041")

toy_graph_weighted = {
    'A': [('B', 2), ('C', 5)],
    'B': [('A', 2), ('D', 4), ('E', 1)],
    'C': [('A', 5), ('F', 2), ('G', 9)],
    'D': [('B', 4), ('G', 3)],
    'E': [('B', 1), ('G', 6)],
    'F': [('C', 2), ('G', 1)],
    'G': [('C', 9), ('D', 3), ('E', 6), ('F', 1)],
}

heuristics = {
    'A': 6,
    'B': 5,
    'C': 2,
    'D': 2,
    'E': 4,
    'F': 1,
    'G': 0
}

print("Toy graph and heuristic values are defined.")
```

```python
print("Laiba Sumaiya Nazim 23549009041")

def path_cost(graph, path):
    total = 0
    for i in range(len(path) - 1):
        current = path[i]
        next_node = path[i + 1]
        for neighbor, cost in graph[current]:
            if neighbor == next_node:
                total += cost
                break
    return total

def reconstruct_path(parent, start, goal):
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent[current]
    path.reverse()
    if path[0] != start:
        return []
    return path

print("Helper functions are ready.")
```

```python
print("Laiba Sumaiya Nazim 23549009041")

def greedy_best_first_search(graph, heuristic, start, goal):
    frontier = []
    heappush(frontier, (heuristic[start], start))

    parent = {start: None}
    visited = set()
    expanded = []

    while frontier:
        h_value, current = heappop(frontier)

        if current in visited:
            continue

        visited.add(current)
        expanded.append(current)

        if current == goal:
            path = reconstruct_path(parent, start, goal)
            return path, path_cost(graph, path), expanded

        for neighbor, cost in graph[current]:
            if neighbor not in visited and neighbor not in parent:
                parent[neighbor] = current
                heappush(frontier, (heuristic[neighbor], neighbor))

    return [], float("inf"), expanded

greedy_path, greedy_cost, greedy_expanded = greedy_best_first_search(
    toy_graph_weighted, heuristics, 'A', 'G'
)

print("Greedy Best-First Search path:", greedy_path)
print("Greedy path cost:", greedy_cost)
print("Greedy expanded nodes:", greedy_expanded)
print("Number of expanded nodes:", len(greedy_expanded))
```

```python
print("Laiba Sumaiya Nazim 23549009041")

def a_star_search(graph, heuristic, start, goal):
    frontier = []
    heappush(frontier, (heuristic[start], 0, start))

    parent = {start: None}
    g_cost = {start: 0}
    expanded = []

    while frontier:
        f_value, current_g, current = heappop(frontier)

        if current_g != g_cost.get(current, float("inf")):
            continue

        expanded.append(current)

        if current == goal:
            path = reconstruct_path(parent, start, goal)
            return path, g_cost[current], expanded

        for neighbor, edge_cost in graph[current]:
            new_g = g_cost[current] + edge_cost

            if new_g < g_cost.get(neighbor, float("inf")):
                g_cost[neighbor] = new_g
                parent[neighbor] = current
                f_value = new_g + heuristic[neighbor]
                heappush(frontier, (f_value, new_g, neighbor))

    return [], float("inf"), expanded

a_star_path, a_star_cost, a_star_expanded = a_star_search(
    toy_graph_weighted, heuristics, 'A', 'G'
)

print("A* Search path:", a_star_path)
print("A* path cost:", a_star_cost)
print("A* expanded nodes:", a_star_expanded)
print("Number of expanded nodes:", len(a_star_expanded))
```

```python
print("Laiba Sumaiya Nazim 23549009041")

print("Toy Graph Comparison")
print("Greedy path:", greedy_path, "Cost:", greedy_cost)
print("A* path:", a_star_path, "Cost:", a_star_cost)

if greedy_path == a_star_path:
    print("Both algorithms found the same path.")
else:
    print("Greedy and A* found different paths.")

print("The heuristic h(B) = 5 makes Greedy prefer B over C because it only looks at h(n).")
print("Greedy can therefore follow A-B-E-G, even though its total cost is higher.")
print("A* also considers the accumulated cost g(n), so it can choose the cheaper route.")
```

```python
print("Laiba Sumaiya Nazim 23549009041")

def true_shortest_costs_to_goal(graph, goal):
    frontier = [(0, goal)]
    distances = {goal: 0}

    while frontier:
        current_cost, current = heappop(frontier)

        if current_cost != distances[current]:
            continue

        for neighbor, edge_cost in graph[current]:
            new_cost = current_cost + edge_cost
            if new_cost < distances.get(neighbor, float("inf")):
                distances[neighbor] = new_cost
                heappush(frontier, (new_cost, neighbor))

    return distances

true_costs = true_shortest_costs_to_goal(toy_graph_weighted, 'G')

print("Admissibility check for the toy graph:")
all_admissible = True

for node in toy_graph_weighted:
    h = heuristics[node]
    true_cost = true_costs[node]
    admissible = h <= true_cost
    if not admissible:
        all_admissible = False
    print(f"{node}: h(n) = {h}, true cost = {true_cost}, admissible = {admissible}")

print("All heuristic values admissible:", all_admissible)
print("An admissible heuristic never overestimates the actual cheapest remaining cost.")
print("This property helps A* guarantee an optimal solution when the usual search conditions are satisfied.")
```

```python
print("Laiba Sumaiya Nazim 23549009041")

def uniform_cost_search(graph, start, goal):
    frontier = [(0, start)]
    parent = {start: None}
    cost_so_far = {start: 0}
    expanded = []

    while frontier:
        current_cost, current = heappop(frontier)

        if current_cost != cost_so_far.get(current, float("inf")):
            continue

        expanded.append(current)

        if current == goal:
            path = reconstruct_path(parent, start, goal)
            return path, current_cost, expanded

        for neighbor, edge_cost in graph[current]:
            new_cost = current_cost + edge_cost

            if new_cost < cost_so_far.get(neighbor, float("inf")):
                cost_so_far[neighbor] = new_cost
                parent[neighbor] = current
                heappush(frontier, (new_cost, neighbor))

    return [], float("inf"), expanded

ucs_path, ucs_cost, ucs_expanded = uniform_cost_search(
    bd_graph_weighted, 'Rangpur', "Cox's Bazar"
)

print("Bangladesh UCS")
print("Path:", ucs_path)
print("Total distance:", ucs_cost)
print("Expanded nodes:", ucs_expanded)
print("Number of expanded nodes:", len(ucs_expanded))

print("\nComparison:")
print("Greedy path:", bd_greedy_path, "Cost:", bd_greedy_cost)
print("A* path:", bd_a_star_path, "Cost:", bd_a_star_cost)
print("UCS path:", ucs_path, "Cost:", ucs_cost)

print("\nDoes A* match UCS?", bd_a_star_path == ucs_path and bd_a_star_cost == ucs_cost)
print("Does Greedy match UCS?", bd_greedy_path == ucs_path and bd_greedy_cost == ucs_cost)
```
