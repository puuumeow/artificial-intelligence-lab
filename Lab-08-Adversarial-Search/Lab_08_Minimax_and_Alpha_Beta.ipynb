```python
game_tree = [
    [[3, 5], [2, 9]],   # A: children C=[3,5], D=[2,9]
    [[1, 4], [8, 3]]    # B: children E=[1,4], F=[8,3]
]

branches = ['A', 'B']
print("Laiba 23549009041")
```

```python
calls = [0]

def minimax(node, is_max):
    calls[0] += 1
    if isinstance(node, int):
        return node
    values = [minimax(child, not is_max) for child in node]
    return max(values) if is_max else min(values)

calls[0] = 0
score = minimax(game_tree, True)

print('Minimax')
print(f'Optimal score at ROOT : {score}')
print(f'Total function calls  : {calls[0]}')
print("Laiba 23549009041")
```

```python
calls[0] = 0
branch_scores = [minimax(child, False) for child in game_tree]
best_idx      = branch_scores.index(max(branch_scores))

print('Best Branch')
print(f'Branch scores  : A={branch_scores[0]}, B={branch_scores[1]}')
print(f'Best branch    : {branches[best_idx]}')
print(f'Expected score : {branch_scores[best_idx]}')
print(f'Function calls : {calls[0]}')
print("Laiba 23549009041")
```

```python
ab_calls = [0]

def alphabeta(node, is_max, alpha=float('-inf'), beta=float('inf')):
    ab_calls[0] += 1
    if isinstance(node, int):
        return node
    if is_max:
        value = float('-inf')
        for child in node:
            value = max(value, alphabeta(child, False, alpha, beta))
            alpha = max(alpha, value)
            if alpha >= beta:
                break                  # beta cutoff
        return value
    else:
        value = float('inf')
        for child in node:
            value = min(value, alphabeta(child, True, alpha, beta))
            beta = min(beta, value)
            if alpha >= beta:
                break                  # alpha cutoff
        return value

ab_calls[0] = 0
alpha_root = float('-inf')            # best score MAX has found so far at the root
ab_scores  = []
for child in game_tree:
    s = alphabeta(child, False, alpha_root, float('inf'))
    ab_scores.append(s)
    alpha_root = max(alpha_root, s)   # pass the best score on to the next branch
best_ab = branches[ab_scores.index(max(ab_scores))]

print('Alpha-Beta Pruning')
print(f'Branch scores  : A={ab_scores[0]}, B={ab_scores[1]}')
print(f'Best branch    : {best_ab}')
print(f'Expected score : {max(ab_scores)}')
print(f'Function calls : {ab_calls[0]}')
print()
print(f'Minimax calls  : {calls[0]}')
print(f'Alpha-beta calls: {ab_calls[0]}')
print(f'Calls saved    : {calls[0] - ab_calls[0]}')
print(f'Call reduction  : {(calls[0] - ab_calls[0]) / calls[0] * 100:.1f}%')
print("Laiba 23549009041")
```

```python
visited = []

def alphabeta_track(node, is_max, alpha=float('-inf'), beta=float('inf')):
    if isinstance(node, int):
        visited.append(node)
        return node
    if is_max:
        value = float('-inf')
        for child in node:
            value = max(value, alphabeta_track(child, False, alpha, beta))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for child in node:
            value = min(value, alphabeta_track(child, True, alpha, beta))
            beta = min(beta, value)
            if alpha >= beta:
                break
        return value

def get_all_leaves(node):
    if isinstance(node, int):
        return [node]
    result = []
    for child in node:
        result.extend(get_all_leaves(child))
    return result

all_leaves = get_all_leaves(game_tree)
visited    = []
alphabeta_track(game_tree, True)

skipped   = len(all_leaves) - len(visited)
reduction = skipped / len(all_leaves) * 100

# record which leaves were not explored
skipped_leaves = all_leaves.copy()
for v in visited:
    skipped_leaves.remove(v)

print('Pruned Leaf Nodes')
print(f'All leaves      : {all_leaves}')
print(f'Visited leaves  : {visited}')
print(f'Skipped leaves  : {skipped_leaves}')
print(f'Leaves evaluated: {len(visited)} / {len(all_leaves)}')
print(f'Leaves skipped  : {skipped}')
print(f'Reduction       : {reduction:.1f}%')
print("Laiba 23549009041")
```

```python
board = [
    ['X', 'O', 'X'],
    ['O', 'X', None],
    [None, None, 'O']
]
print("Laiba 23549009041")
```

```python
LINES = [
    [(0,0),(0,1),(0,2)], [(1,0),(1,1),(1,2)], [(2,0),(2,1),(2,2)],   # rows
    [(0,0),(1,0),(2,0)], [(0,1),(1,1),(2,1)], [(0,2),(1,2),(2,2)],   # columns
    [(0,0),(1,1),(2,2)], [(0,2),(1,1),(2,0)]                          # diagonals
]

def evaluate(board):
    for line in LINES:
        a, b, c = [board[r][col] for r, col in line]
        if a is not None and a == b == c:
            return a + ' wins'
    if all(cell is not None for row in board for cell in row):
        return 'Draw'
    return 'Ongoing'

def show(board):
    for row in board:
        print(' | '.join(cell if cell else ' ' for cell in row))

print('Board:')
show(board)
print('Result:', evaluate(board))
print("Laiba 23549009041")
```

```python
def get_moves(board):
    return [(r, c) for r in range(3) for c in range(3) if board[r][c] is None]

calls = [0]

def minimax_ttt(board, is_max):
    calls[0] += 1
    result = evaluate(board)
    if result == 'X wins': return 1
    if result == 'O wins': return -1
    if result == 'Draw':   return 0
    player = 'X' if is_max else 'O'
    scores = []
    for (r, c) in get_moves(board):
        board[r][c] = player                       # try the move
        scores.append(minimax_ttt(board, not is_max))
        board[r][c] = None                         # undo the move
    return max(scores) if is_max else min(scores)

calls[0] = 0
score = minimax_ttt(board, True)

print('Minimax')
print(f'Optimal score  : {score}   (1 = X wins, 0 = draw, -1 = O wins)')
print(f'Function calls : {calls[0]}')
print("Laiba 23549009041")
```

```python
def best_move_minimax(board):
    calls[0] = 1                                   # counts the starting board itself
    move_scores = {}
    for (r, c) in get_moves(board):
        board[r][c] = 'X'
        move_scores[(r, c)] = minimax_ttt(board, False)
        board[r][c] = None
    best = max(move_scores, key=move_scores.get)
    return best, move_scores

best_move, move_scores = best_move_minimax(board)
q3_calls = calls[0]

print('Best Move (Minimax)')
for mv, sc in move_scores.items():
    print(f'  X plays {mv} -> score {sc}')
print(f'Best move      : {best_move}')
print(f'Expected score : {move_scores[best_move]}')
print(f'Function calls : {q3_calls}')
print("Laiba 23549009041")
```

```python
skipped_moves = []
track_calls = [0]

def alphabeta_track(board, is_max, path, alpha=float('-inf'), beta=float('inf')):
    track_calls[0] += 1
    result = evaluate(board)
    if result == 'X wins': return 1
    if result == 'O wins': return -1
    if result == 'Draw':   return 0
    player = 'X' if is_max else 'O'
    moves = get_moves(board)
    value = float('-inf') if is_max else float('inf')
    for i, (r, c) in enumerate(moves):
        board[r][c] = player
        child = alphabeta_track(board, not is_max, path + [f'{player}{(r, c)}'], alpha, beta)
        board[r][c] = None
        if is_max:
            value = max(value, child)
            alpha = max(alpha, value)
        else:
            value = min(value, child)
            beta = min(beta, value)
        if alpha >= beta:
            for (r2, c2) in moves[i + 1:]:     # these moves are never explored
                skipped_moves.append(path + [f'{player}{(r2, c2)}'])
            break
    return value

skipped_moves = []
track_calls[0] = 1
alpha = float('-inf')
for (r, c) in get_moves(board):
    board[r][c] = 'X'
    s = alphabeta_track(board, False, [f'X{(r, c)}'], alpha, float('inf'))
    board[r][c] = None
    alpha = max(alpha, s)

reduction = (q3_calls - track_calls[0]) / q3_calls * 100

print('Moves NOT explored (pruned):')
for path in skipped_moves:
    print('  ' + '  ->  '.join(path))
print(f'Number of pruned moves : {len(skipped_moves)}')
print()
print(f'Minimax calls    : {q3_calls}')
print(f'Alpha-beta calls : {track_calls[0]}')
print(f'Calls saved      : {q3_calls - track_calls[0]}')
print(f'Reduction        : {reduction:.1f}%')
print("Laiba 23549009041")
```
