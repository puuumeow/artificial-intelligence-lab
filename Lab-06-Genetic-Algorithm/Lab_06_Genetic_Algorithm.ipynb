```python
import random
CHROMOSOME_LENGTH = 5
POP_SIZE = 30
MUTATION_RATE = 0.05
GENERATIONS = 60
TOURNAMENT_K = 3
def decode(chromosome):
    return int(chromosome, 2)
def f(x):
    return x ** 2
def create_individual():
    return "".join(random.choice("01") for _ in range(CHROMOSOME_LENGTH))
print("Random individual:", create_individual())
print("Laiba 23549009041")
```

```python
def fitness(individual):
    x = decode(individual)
    return f(x)
test_individual = "10110"
print("Chromosome:", test_individual)
print("Decoded value:", decode(test_individual))
print("Fitness:", fitness(test_individual))
print("Laiba 23549009041")
```

```python
def select_parent(population, fitnesses):
    indices = random.choices(
        range(len(population)),
        k=TOURNAMENT_K
    )
    best_index = max(
        indices,
        key=lambda i: fitnesses[i]
    )
    return population[best_index]
population = [create_individual() for _ in range(POP_SIZE)]
fitnesses = [fitness(ind) for ind in population]
parent = select_parent(population, fitnesses)
print("Selected parent:", parent)
print("Parent value:", decode(parent))
print("Parent fitness:", fitness(parent))
print("Laiba 23549009041")
```

```python
def crossover(parent1, parent2):
    cut = random.randint(1, CHROMOSOME_LENGTH - 1)
    child = parent1[:cut] + parent2[cut:]
    return child
def mutate(individual):
    mutated = []
    for bit in individual:
        if random.random() < MUTATION_RATE:
            if bit == "0":
                mutated.append("1")
            else:
                mutated.append("0")
        else:
            mutated.append(bit)

    return "".join(mutated)
parent1 = create_individual()
parent2 = create_individual()
child = crossover(parent1, parent2)
mutated_child = mutate(child)
print("Parent 1:", parent1)
print("Parent 2:", parent2)
print("Child:", child)
print("After mutation:", mutated_child)
print("Laiba 23549009041")
```

```python
def run_ga():
    population = [
        create_individual()
        for _ in range(POP_SIZE)
    ]
    best_ind, best_fit = None, -1
    for gen in range(GENERATIONS):
        fitnesses = [
            fitness(ind)
            for ind in population
        ]
        gen_best_idx = fitnesses.index(
            max(fitnesses)
        )
        gen_best_ind = population[gen_best_idx]
        gen_best_fit = fitnesses[gen_best_idx]
        if gen_best_fit > best_fit:
            best_ind = gen_best_ind
            best_fit = gen_best_fit
        if gen % 10 == 0:
            print(
                f"Gen {gen:3d} | "
                f"best so far: "
                f"x={decode(best_ind)} "
                f"chromosome={best_ind} "
                f"f(x)={best_fit}"
            )
        new_population = [gen_best_ind]
        while len(new_population) < POP_SIZE:

            # Selection
            p1 = select_parent(
                population,
                fitnesses
            )
            p2 = select_parent(
                population,
                fitnesses
            )
            child = crossover(p1, p2)
            child = mutate(child)
            new_population.append(child)
        population = new_population
    print()
    print(
        f"Final answer: "
        f"x = {decode(best_ind)} "
        f"(f(x) = {best_fit})"
    )
    print(
        "Brute-force check: "
        "the true maximum on [0,31] "
        "is x=31, f(x)=961."
    )
run_ga()
print("Laiba 23549009041")
```
