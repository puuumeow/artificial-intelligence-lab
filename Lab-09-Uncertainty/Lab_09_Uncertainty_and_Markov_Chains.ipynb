# AI Lab 9: Uncertainty, Markov Chains and Propositional Logic

This notebook completes the supplied lab tasks for **Uncertainty** and **Markov Chains and Language Models**. The weather probabilities are illustrative, while the breast-cancer records are real data included with scikit-learn.

> Note: The supplied lab questions do not include a separate propositional-logic exercise, so this notebook covers the two topics and tasks explicitly provided.

```python
print("laiba sumaiya nazim 23549009041")
import numpy as np
import pandas as pd

# The three possible weather states, in a fixed order.
states = ['Sunny', 'Cloudy', 'Rainy']

# Initial probability distribution: P(today).
init = {'Sunny': 0.5, 'Cloudy': 0.3, 'Rainy': 0.2}

# Transition probabilities: P(tomorrow | today).
# Each row represents today's state; each inner value represents tomorrow's state.
T = {
    'Sunny':  {'Sunny': 0.6, 'Cloudy': 0.3, 'Rainy': 0.1},
    'Cloudy': {'Sunny': 0.3, 'Cloudy': 0.4, 'Rainy': 0.3},
    'Rainy':  {'Sunny': 0.2, 'Cloudy': 0.4, 'Rainy': 0.4},
}
```

### Task 1.1 — Joint distribution

The joint probability is calculated using:

\[
P(\text{today},\text{tomorrow}) =
P(\text{today})P(\text{tomorrow}\mid\text{today})
\]

Rows represent today's weather and columns represent tomorrow's weather. All nine values must sum to 1 because they cover every possible pair of states.

```python
print("laiba sumaiya nazim 23549009041")
# Build the 3 x 3 joint probability table.
joint = pd.DataFrame(
    [[init[t] * T[t][u] for u in states] for t in states],
    index=states, columns=states
)
joint.index.name = 'today'
joint.columns.name = 'tomorrow'

print("Joint P(today, tomorrow):")
print(joint.round(3))
print("\nSum of all 9 entries:", joint.values.sum())
assert np.isclose(joint.values.sum(), 1.0), "Joint probabilities must sum to 1."
```

### Task 1.2 — Marginal probabilities

A marginal probability is found by summing the joint probabilities over the other variable.

- Sum each column to obtain \(P(\text{tomorrow})\).
- Sum each row to obtain \(P(\text{today})\).

```python
print("laiba sumaiya nazim 23549009041")
# Sum down each column to get P(tomorrow).
p_tomorrow = joint.sum(axis=0)

# Sum across each row to get P(today).
p_today = joint.sum(axis=1)

print("P(tomorrow):")
print(p_tomorrow.round(3))
print("\nP(today), checked from the joint table:")
print(p_today.round(3))
print("\nSum of tomorrow's marginal:", p_tomorrow.sum())
```

### Task 1.3 — Conditional probabilities

Conditional probability is calculated by dividing a joint probability by the probability of the condition.

\[
P(B\mid A)=\frac{P(A,B)}{P(A)}
\]

The first calculation finds tomorrow's distribution when today is Rainy. The second finds today's distribution when tomorrow is Rainy. Both are calculated from the joint table.

```python
print("laiba sumaiya nazim 23549009041")
# P(tomorrow | today = Rainy): divide the Rainy row by its row sum.
p_tom_given_rainy = joint.loc['Rainy'] / joint.loc['Rainy'].sum()

# P(today | tomorrow = Rainy): divide the Rainy column by its column sum.
p_today_given_rainy = joint['Rainy'] / joint['Rainy'].sum()

print("P(tomorrow | today = Rainy):")
print(p_tom_given_rainy.round(3))
print("\nP(today | tomorrow = Rainy):")
print(p_today_given_rainy.round(3))
```

### Task 1.4 — Bayes' rule

Bayes' rule is used to reverse a conditional probability:

\[
P(A\mid B)=\frac{P(B\mid A)P(A)}{P(B)}
\]

Here, it calculates the probability that today was Rainy given that tomorrow is Rainy. The result is compared with the direct conditional probability calculated from the joint table.

```python
print("laiba sumaiya nazim 23549009041")
# Likelihood: P(tomorrow = Rainy | today = Rainy), derived from the joint table.
likelihood = joint.loc['Rainy', 'Rainy'] / p_today['Rainy']
prior = p_today['Rainy']
evidence = p_tomorrow['Rainy']

# Apply Bayes' rule.
bayes = likelihood * prior / evidence

# Compare with the direct conditional probability from the joint table.
direct = p_today_given_rainy['Rainy']

print(f"Likelihood P(tomorrow=Rainy | today=Rainy) = {likelihood:.4f}")
print(f"Prior P(today=Rainy) = {prior:.4f}")
print(f"Evidence P(tomorrow=Rainy) = {evidence:.4f}")
print(f"Bayes result P(today=Rainy | tomorrow=Rainy) = {bayes:.4f}")
print(f"Direct result from joint table = {direct:.4f}")
print("Results match:", np.isclose(bayes, direct))
```

### Task 1.5 — Independence

Two events are independent if:

\[
P(A,B)=P(A)P(B)
\]

Compare the joint probability of Sunny on both days with the product of the two marginal probabilities. If the values are different, the events are dependent.

```python
print("laiba sumaiya nazim 23549009041")
p_joint_ss = joint.loc['Sunny', 'Sunny']
p_product = p_today['Sunny'] * p_tomorrow['Sunny']

print(f"P(Sunny, Sunny)               = {p_joint_ss:.4f}")
print(f"P(today=Sunny) x P(tom=Sunny) = {p_product:.4f}")
print("Independent?", np.isclose(p_joint_ss, p_product))
# Not equal, so today's and tomorrow's weather are dependent.
```

# Topic 2: Markov Chains and Language Model

A Markov chain is a model in which the next state depends on the current state. Its transition matrix stores the probabilities of moving between states.

## Part A — Weather Markov Chain

The transition matrix uses the same state order for its rows and columns: Sunny, Cloudy, Rainy. A probability distribution is represented as a row vector and is multiplied on the right by the transition matrix.

### A.1 — Construct the transition matrix

Convert the dictionary `T` into a NumPy matrix. Each row must sum to 1 because it lists all possible next-weather probabilities for one current state.

```python
print("laiba sumaiya nazim 23549009041")
P = np.array([[T[s][t] for t in states] for s in states])
print(P)
print("Row sums:", P.sum(axis=1))
```

### A.2 — Simulate 10 days starting from Sunny

At each step, `rng.choice` selects the next state using the probability distribution in the current state's row. The fixed random seed makes the sample sequence reproducible.

```python
print("laiba sumaiya nazim 23549009041")
rng = np.random.default_rng(42)   # remove the seed for different runs

def simulate(start, n_days):
    seq = [start]
    for _ in range(n_days - 1):
        i = states.index(seq[-1])
        seq.append(rng.choice(states, p=P[i]))
    return seq

print(" -> ".join(simulate('Sunny', 10)))
```

### A.3 — Distribution after 2 and 30 days

For a starting row-vector distribution \(v_0\), the distribution after \(n\) transitions is:

\[
v_n=v_0P^n
\]

`matrix_power(P, n)` calculates the matrix power efficiently.

```python
print("laiba sumaiya nazim 23549009041")
start = np.array([1.0, 0.0, 0.0])   # Sunny

dist_2 = start @ np.linalg.matrix_power(P, 2)
dist_30 = start @ np.linalg.matrix_power(P, 30)

print("After 2 days: ", dict(zip(states, dist_2.round(4))))
print("After 30 days:", dict(zip(states, dist_30.round(4))))
```

### A.4 — Multiply the initial distribution by P 100 times

This repeatedly applies one transition. Starting once from Sunny and once from Rainy shows whether the chain approaches the same long-run (stationary) distribution.

```python
print("laiba sumaiya nazim 23549009041")
for start in ([1, 0, 0], [0, 0, 1]):
    v = np.array(start, dtype=float)
    for _ in range(100):
        v = v @ P
    print(f"Start {start} -> {v.round(4)}")

# Both starts give the same result: the stationary distribution
```

# Topic 1 — Task 2: Real Data (Breast Cancer Diagnostic Dataset)

The breast-cancer dataset is built into scikit-learn and contains 569 records. The `target` value is 0 for malignant and 1 for benign. We create three Boolean columns: `malignant`, `large` (mean radius above its median), and `rough` (mean texture above its median).

```python
print("laiba sumaiya nazim 23549009041")
from sklearn.datasets import load_breast_cancer

# Load the built-in dataset as a pandas DataFrame.
df = load_breast_cancer(as_frame=True).frame

# Create Boolean features according to the lab instructions.
df['malignant'] = (df['target'] == 0)
df['large'] = df['mean radius'] > df['mean radius'].median()
df['rough'] = df['mean texture'] > df['mean texture'].median()

print("DataFrame shape:", df.shape)
print("\nTumor counts:")
print(df['malignant'].map({True: 'Malignant', False: 'Benign'}).value_counts())
print("\nMalignant count:", int(df['malignant'].sum()))
print("Benign count:", int((~df['malignant']).sum()))
```

### Task 2.1 — Joint distribution of malignant and large

`pd.crosstab(..., normalize='all')` calculates each cell's fraction of the complete dataset. The four cells should sum to 1.

```python
print("laiba sumaiya nazim 23549009041")
joint_cancer = pd.crosstab(df['malignant'], df['large'], normalize='all')
joint_cancer.index = ['Benign' if not x else 'Malignant' for x in joint_cancer.index]
joint_cancer.columns = ['Not large' if not x else 'Large' for x in joint_cancer.columns]

print("Joint distribution P(malignant status, size):")
print(joint_cancer)
print("\nSum of all joint probabilities:", joint_cancer.to_numpy().sum())
assert np.isclose(joint_cancer.to_numpy().sum(), 1.0)
```

### Task 2.2 — Marginal and conditional probabilities

The marginal probabilities are obtained by summing rows or columns. Conditional probabilities are calculated by dividing the relevant joint probability by the probability of the condition.

```python
print("laiba sumaiya nazim 23549009041")
# Use the original Boolean-indexed crosstab for transparent calculations.
joint_bool = pd.crosstab(df['malignant'], df['large'], normalize='all')

p_malignant = df['malignant'].mean()
p_large = df['large'].mean()
p_large_given_malignant = joint_bool.loc[True, True] / p_malignant
p_large_given_benign = joint_bool.loc[False, True] / (1 - p_malignant)

print(f"P(malignant) = {p_malignant:.4f}")
print(f"P(large) = {p_large:.4f}")
print(f"P(large | malignant) = {p_large_given_malignant:.4f}")
print(f"P(large | benign) = {p_large_given_benign:.4f}")
```

### Task 2.3 — Bayes' rule and direct verification

Use Bayes' rule to calculate \(P(\text{malignant}\mid\text{large})\), then verify it by filtering the original records. Also calculate the probability of malignancy among tumors that are not large.

```python
print("laiba sumaiya nazim 23549009041")
# Bayes' rule: P(malignant | large)
p_malignant_given_large_bayes = (
    p_large_given_malignant * p_malignant / p_large
)

# Direct count among records classified as large.
large_records = df[df['large']]
p_malignant_given_large_direct = large_records['malignant'].mean()

# Probability of malignancy among records that are not large.
not_large_records = df[~df['large']]
p_malignant_given_not_large = not_large_records['malignant'].mean()

print(f"Bayes P(malignant | large) = {p_malignant_given_large_bayes:.4f}")
print(f"Direct P(malignant | large) = {p_malignant_given_large_direct:.4f}")
print("Bayes and direct results match:",
      np.isclose(p_malignant_given_large_bayes, p_malignant_given_large_direct))
print(f"P(malignant | not large) = {p_malignant_given_not_large:.4f}")
```

### Task 2.4 — Independence

Compare \(P(\text{malignant},\text{large})\) with \(P(\text{malignant})P(\text{large})\). If these values are not equal, malignancy and the large-radius indicator are not independent in this dataset.

```python
print("laiba sumaiya nazim 23549009041")
p_malignant_and_large = joint_bool.loc[True, True]
p_malignant_times_large = p_malignant * p_large

print(f"P(malignant, large) = {p_malignant_and_large:.4f}")
print(f"P(malignant) x P(large) = {p_malignant_times_large:.4f}")
print("Independent?", np.isclose(p_malignant_and_large, p_malignant_times_large))
```

## Topic 2 — Part B: Word Markov Chain (Bigram Language Model)

In this part, each word is treated as a state. A bigram is a pair of neighboring tokens. The special token `<s>` marks the start of a sentence and `</s>` marks its end. The transition probability is:

\[
P(w_{i+1}\mid w_i)=\frac{\operatorname{count}(w_i,w_{i+1})}
{\operatorname{count}(w_i)}
\]

The model is trained only on the four supplied sentences.

### B.1 — Count words and neighboring pairs

The sentences are tokenized into words, then `<s>` and `</s>` are added. `Counter` counts individual tokens and adjacent token pairs.

```python
print("laiba sumaiya nazim 23549009041")
from collections import Counter
import re

corpus = [
    "the student writes the code.",
    "the student runs the code.",
    "the code prints the result.",
    "the teacher checks the code.",
]

tokenized_sentences = []
word_counts = Counter()
bigram_counts = Counter()

for sentence in corpus:
    # Extract lowercase alphabetic words; punctuation is not treated as a state.
    words = re.findall(r"[a-z]+", sentence.lower())
    tokens = ['<s>'] + words + ['</s>']
    tokenized_sentences.append(tokens)

    word_counts.update(tokens)
    bigram_counts.update(zip(tokens[:-1], tokens[1:]))

print("Tokenized sentences:")
for tokens in tokenized_sentences:
    print(tokens)

print("\nWord counts:")
print(dict(word_counts))

print("\nNeighboring-word pair counts:")
for pair, count in sorted(bigram_counts.items()):
    print(f"{pair}: {count}")
```

### B.2 — Calculate transition probabilities

For every observed pair, divide its count by the count of the current word. Then display the possible next words after `the`.

```python
print("laiba sumaiya nazim 23549009041")
bigram_probabilities = {
    pair: count / word_counts[pair[0]]
    for pair, count in bigram_counts.items()
}

print("Probabilities of words that can follow 'the':")
for (current, next_word), probability in sorted(bigram_probabilities.items()):
    if current == 'the':
        print(f"P({next_word} | the) = {probability:.4f}")
```

### B.3 — Probability of complete word sequences

The probability of a sentence is the product of the transition probabilities along its path, including the start and end markers. If any required transition was never observed, its probability is zero.

```python
print("laiba sumaiya nazim 23549009041")
def sentence_probability(sentence):
    words = re.findall(r"[a-z]+", sentence.lower())
    tokens = ['<s>'] + words + ['</s>']
    probability = 1.0

    for current, next_word in zip(tokens[:-1], tokens[1:]):
        probability *= bigram_probabilities.get((current, next_word), 0.0)

    return probability

sentence_a = "the student runs the code."
sentence_b = "the teacher runs the code."

print(f"P({sentence_a!r}) = {sentence_probability(sentence_a):.8f}")
print(f"P({sentence_b!r}) = {sentence_probability(sentence_b):.8f}")
```

### B.4 — Generate three sentences

Start at `<s>`, sample a next token according to the transition probabilities for the current token, and continue until `</s>` is selected. A maximum-length guard prevents an accidental infinite loop. A fixed random seed makes the three generated samples reproducible.

```python
print("laiba sumaiya nazim 23549009041")
rng_words = np.random.default_rng(42)

# Build a list of possible next words and their probabilities for each current word.
next_word_options = {}
for current in word_counts:
    options = [
        (next_word, probability)
        for (source, next_word), probability in bigram_probabilities.items()
        if source == current
    ]
    if options:
        next_word_options[current] = options

def generate_sentence(max_tokens=30):
    current = '<s>'
    generated = []

    for _ in range(max_tokens):
        options = next_word_options.get(current)
        if not options:
            break

        next_words = [word for word, _ in options]
        probabilities = np.array([prob for _, prob in options], dtype=float)
        probabilities = probabilities / probabilities.sum()

        next_word = rng_words.choice(next_words, p=probabilities)
        if next_word == '</s>':
            break

        generated.append(next_word)
        current = next_word

    return ' '.join(generated)

print("Three generated sentences:")
for i in range(3):
    print(f"{i + 1}. {generate_sentence()}")
```
