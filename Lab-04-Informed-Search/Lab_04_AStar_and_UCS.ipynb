This notebook implements graph search algorithms for both a toy graph and a Bangladesh weighted graph. The algorithms include A* Search and Uniform Cost Search. The notebook also shows the tracing of the algorithms and checks heuristic admissibility and consistency.

```python
sabcg_graph = {

  'S': [('A', 1), ('B', 1)],
  'A': [('C', 1)],
  'B': [('C', 2)],
  'C': [('G', 3)],
  'G': [],
}
h_sabcg = {
    'S': 2 , 'A' : 4 , 'B' : 1 , 'C' : 1 , 'G' : 0
}
```

```python
def astar_priority_sabcg(g, node):
  return g + h_sabcg[node]
```

```python
def graph_search(graph, start, goal, priority_fn):
  closed = []
  frontier = [(priority_fn(0, start), 0, start , [start])]
  expansions = 0

  while len(frontier) > 0:
    best_index = 0
    for i in range(len(frontier)):
      if frontier[i][0] < frontier[best_index][0]:
        best_index = i
    priority , g , node, path = frontier.pop(best_index)


    if node == goal :
      return path , g ,expansions

    if node not in closed :
      closed.append(node)
      expansions = expansions +1
      for neighbore , weight in graph[node]:
        new_g = g + weight
        new_path = path + [neighbore] # Corrected line
        frontier.append((priority_fn(new_g, neighbore),new_g, neighbore, new_path))
  return None, None, expansions
```

A* trace function

```python
print("Laiba 23549009041")

def trace_astar(graph, start, goal):

    closed = []

    frontier = [
        (
            astar_priority_bd(0, start),
            0,
            start,
            [start]
        )
    ]

    step = 0

    print(
        "Step | Node | g | h | f = g + h"
    )

    print("-" * 45)

    while len(frontier) > 0:

        best_index = 0

        for i in range(len(frontier)):

            if frontier[i][0] < frontier[best_index][0]:

                best_index = i

        priority, g, node, path = frontier.pop(best_index)

        step = step + 1

        print(
            f"{step} | {node} | {g} | "
            f"{bd_heuristics[node]} | {priority}"
        )

        if node == goal:

            print("\nGoal found!")

            return path, g

        if node not in closed:

            closed.append(node)

            for neighbor, weight in graph[node]:

                new_g = g + weight

                new_path = path + [neighbor]

                new_f = astar_priority_bd(
                    new_g,
                    neighbor
                )

                frontier.append(
                    (
                        new_f,
                        new_g,
                        neighbor,
                        new_path
                    )
                )

    return None, None
```

Run A* Trace

```python
print("Laiba 23549009041")

trace_path, trace_cost = trace_astar(
    bd_graph_weighted,
    'Rangpur',
    "Cox's Bazar"
)

print("\nFinal Path:", trace_path)
print("Final Cost:", trace_cost)
```

```python
print("Laiba 23549009041")

def trace_ucs(graph, start, goal):

    closed = []

    frontier = [
        (
            ucs_priority(0, start),
            0,
            start,
            [start]
        )
    ]

    step = 0

    print(
        "Step | Node | g | h | f = g + h"
    )

    print("-" * 45)

    while len(frontier) > 0:

        best_index = 0

        for i in range(len(frontier)):

            if frontier[i][0] < frontier[best_index][0]:

                best_index = i

        priority, g, node, path = frontier.pop(best_index)

        step = step + 1

        h = bd_heuristics[node]

        f = g + h

        print(
            f"{step} | {node} | {g} | {h} | {f}"
        )

        if node == goal:

            print("\nGoal found!")

            return path, g

        if node not in closed:

            closed.append(node)

            for neighbor, weight in graph[node]:

                new_g = g + weight

                new_path = path + [neighbor]

                frontier.append(
                    (
                        new_g,
                        new_g,
                        neighbor,
                        new_path
                    )
                )

    return None, None
```

Check admissibility

```python
print("Laiba 23549009041")

print("ADMISSIBILITY CHECK")
print("-" * 60)

for node in bd_heuristics:

    h = bd_heuristics[node]

    true_cost = true_cost_bd[node]

    if h <= true_cost:

        print(
            f"{node}: Admissible "
            f"because h = {h} <= "
            f"true cost = {true_cost}"
        )

    else:

        print(
            f"{node}: NOT ADMISSIBLE "
            f"because h = {h} > "
            f"true cost = {true_cost}"
        )
```

```python
print("Laiba 23549009041")

print("CONSISTENCY CHECK")
print("-" * 70)

violation_found = False

for node in bd_graph_weighted:

    for neighbor, cost in bd_graph_weighted[node]:

        left = bd_heuristics[node]

        right = cost + bd_heuristics[neighbor]

        if left > right:

            violation_found = True

            print(
                f"VIOLATION: {node} -> {neighbor}"
            )

            print(
                f"h({node}) = {left} > "
                f"{cost} + {bd_heuristics[neighbor]} = {right}"
            )

            print()

if violation_found == False:

    print("No consistency violation found.")
```
