```python
neighbors = {
    'A': ['B'],
    'B': ['A', 'D', 'E', 'F'],
    'C': ['D', 'E'],
    'D': ['B', 'C', 'E', 'F'],
    'E': ['B', 'C', 'D'],
    'F': ['B', 'D'],
}

domains = {v: ['S1', 'S2', 'S3'] for v in 'ABCDE'}
domains['F'] = ['S2', 'S3']

variables = ['A', 'B', 'C', 'D', 'E', 'F']
slots    = ['S1', 'S2', 'S3']
```

```python
#Forward Checking
print("=== Forward Checking ===")
assignment = {}

for var in variables:
    placed = False
    for val in domains[var]:
        if all(assignment.get(n) != val for n in neighbors[var]):
            assignment[var] = val
            print(f"  {var} = {val}")
            placed = True
            break
    if not placed:
        used = {n: assignment[n] for n in neighbors[var] if n in assignment}
        print(f"  STUCK at {var}. Neighbors already assigned: {used}")
        break
```

```python
attempts   = [0]
backtracks = [0]

def is_valid(var, val, assignment):
    return all(assignment.get(n) != val for n in neighbors[var])

def backtrack(assignment):
    if len(assignment) == len(variables):
        return assignment
    var = variables[len(assignment)]
    for val in domains[var]:
        attempts[0] += 1
        if is_valid(var, val, assignment):
            assignment[var] = val
            result = backtrack(assignment)
            if result:
                return result
            del assignment[var]
    backtracks[0] += 1
    return None

solution = backtrack({})

print("\n=== Backtracking ===")
print(f"Schedule:  {solution}")
print(f"Attempts:  {attempts[0]}")
print(f"Backtracks:{backtracks[0]}")
```
